```python
import pandas as pd

# Messy employee data (numbers stored as text)
df = pd.DataFrame({
    "emp_id": ["101", "102", "103", "104"],
    "name": ["Ravi", "Priya", "Arun", "Divya"],
    "age": ["25", "30", "abc", "28"],
    "salary": ["50,000", "62000", "58,500", "N/A"],
    "join_date": ["2022-01-15", "2021-06-10", "2023-03-01", "2020-11-25"],
    "department": ["IT", "HR", "IT", "Sales"],
    "is_active": ["True", "False", "True", "True"],
})

# 1. Original dtypes
print(df.dtypes)

# 2. Conversions
df["emp_id"] = df["emp_id"].astype(int)
df["age"] = pd.to_numeric(df["age"], errors="coerce")
df["salary"] = pd.to_numeric(df["salary"].str.replace(",", ""), errors="coerce")
df["join_date"] = pd.to_datetime(df["join_date"])
df["department"] = df["department"].astype("category")
df["is_active"] = df["is_active"].map({"True": True, "False": False})

# 3. Corrected dtypes
print(df.dtypes)
df.to_csv("employee_cleaned.csv", index=False)
```
